```python
import pandas as pd

# --- Caminho do arquivo ---
df = pd.read_csv("UDS.csv", low_memory=False)

# --- Total de linhas e colunas ---
n_linhas, n_colunas = df.shape
print(f"📊 O dataset possui {n_linhas:,} linhas e {n_colunas:,} colunas.")

# --- Quantidade de participantes únicos (NACCID) ---
if 'NACCID' in df.columns:
    n_pessoas = df['NACCID'].nunique()
    print(f"👥 Total de participantes com dados (NACCID únicos): {n_pessoas:,}")
else:
    print("⚠️ A coluna 'NACCID' não foi encontrada no dataset.")
```
